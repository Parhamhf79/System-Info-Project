#Step 2: Import Library

```python
import streq as sq
```

#*Example Sequence used for this project: ATCCCACCTGCATGGATCCGATCAAAGAAGACTACTATCG*

#Step 3: Transforming a sequence to its Reverse complement

```python
sq.reverse_complement('ATCCCACCTGCATGGATCCGATCAAAGAAGACTACTATCG')
```

#Step 4: Convert a sequence into RNA

```python
sq.to_rna('ATCCCACCTGCATGGATCCGATCAAAGAAGACTACTATCG')
```

#Step 5: calculating GC content of a sequence

```python
sq.gc_content('ATCCCACCTGCATGGATCCGATCAAAGAAGACTACTATCG')
```

#Step 6: Find common Type IIS restriction sites in a sequence

```python
sq.which_re_sites('ATCCCACCTGCATGGATCCGATCAAAGAAGACTACTATCG')
```
